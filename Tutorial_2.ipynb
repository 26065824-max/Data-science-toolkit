```python
import numpy as np

#question 1
array_2x2=np.array([[1,2],[3,4]])
print(array_2x2)
print("Shape: " ,array_2x2.shape)


#question2
random_array = np.random.rand(3, 4)

print(random_array)
print("Shape:", random_array.shape)

#question3
ones_array = np.ones((3, 3))
zeros_array = np.zeros((3, 3))

print("Array of ones:")
print(ones_array)

print("\nArray of zeros:")
print(zeros_array)

#question4
random_array = np.random.rand(6, 3)

transposed_array = random_array.T

print("Original array (6 x 3):")
print(random_array)
print("Original shape:", random_array.shape)

print("\nTransposed array (3 x 6):")
print(transposed_array)
print("Transposed shape:", transposed_array.shape)


#question5
import numpy as np

n = 2
p = 3
m = 4

array_A = np.random.rand(n, p)
array_B = np.random.rand(p, m)

dot_product = np.dot(array_A, array_B)

print("Array A (n x p):")
print(array_A)
print("Shape of A:", array_A.shape)

print("\nArray B (p x m):")
print(array_B)
print("Shape of B:", array_B.shape)

print("\nDot product (n x m):")
print(dot_product)
print("Shape of result:", dot_product.shape)
```
